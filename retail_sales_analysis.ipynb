```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("train[1].csv")

print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

df = df.dropna()

print("\nDuplicate Records:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nCleaned Data:")
print(df.head())
```

```python
high_sales = df[df['Sales'] > 500]

print(high_sales)
```

```python
technology = df[df['Category'] == 'Technology']

print(technology)
```

```python
sorted_sales = df.sort_values(
    by='Sales',
    ascending=False
)

print(sorted_sales.head(10))
```

```python
print("Total Sales:", df['Sales'].sum())

print("Average Sales:", df['Sales'].mean())

print("Maximum Sales:", df['Sales'].max())

print("Minimum Sales:", df['Sales'].min())

print("Sales Standard Deviation:", df['Sales'].std())
```

```python
region_sales = df.groupby('Region')['Sales'].agg([
    'sum',
    'mean',
    'count'
])

print(region_sales)
```

```python
category_sales = df.groupby('Category')['Sales'].agg([
    'sum',
    'mean',
    'count'
])

print(category_sales)
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(8, 5))

plt.bar(
    region_sales.index,
    region_sales.values
)

plt.title("Region-wise Sales")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df['Sales'],
    bins=30,
    edgecolor='black'
)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['Sales'],
    df['Postal Code']
)

plt.title("Sales vs Postal Code")
plt.xlabel("Sales")
plt.ylabel("Postal Code")

plt.show()
```
